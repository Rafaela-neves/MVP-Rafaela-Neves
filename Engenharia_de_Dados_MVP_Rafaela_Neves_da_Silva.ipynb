# **TRATAMENTO**

```python
import numpy as np

# Substituir 'NaN ' (texto) por NaN real em TODAS as colunas
df = df.replace('NaN ', np.nan)
```

```python
# Converter colunas numéricas
df['Delivery_person_Age'] = pd.to_numeric(df['Delivery_person_Age'])
df['Delivery_person_Ratings'] = pd.to_numeric(df['Delivery_person_Ratings'])
```

```python
# Garantir que Time_taken(min) seja string antes de limpar
df['Time_taken(min)'] = df['Time_taken(min)'].astype(str)

# Remover texto e converter para inteiro
df['Time_taken(min)'] = (
    df['Time_taken(min)']
    .str.replace('(min)', '', regex=False)
    .str.strip()
    .astype(int)
)
```

```python
df[['Delivery_person_Age',
    'Delivery_person_Ratings',
    'Time_taken(min)']].describe()
```

```python
df['City'].unique()
```

```python
# Padronizar City definitivamente
df['City'] = df['City'].str.strip()
```

```python
# Tratar idades inválidas
df.loc[df['Delivery_person_Age'] < 18, 'Delivery_person_Age'] = pd.NA
```

```python
# Tratar ratings fora do intervalo
df.loc[df['Delivery_person_Ratings'] > 5, 'Delivery_person_Ratings'] = pd.NA
```

A análise de qualidade dos dados indicou que o conjunto não apresenta valores ausentes originalmente. No entanto, foram identificadas inconsistências semânticas e estruturais em atributos numéricos e categóricos.

As colunas Delivery_person_Age e Delivery_person_Ratings encontravam-se inicialmente armazenadas como texto, exigindo conversão de tipos. Durante esse processo, foram identificados valores inválidos, como idades inferiores a 18 anos e avaliações acima do limite máximo esperado (5), os quais foram tratados como valores ausentes.

Adicionalmente, a variável categórica City apresentou inconsistências de padronização textual, como espaços em branco e valores representando ausência de informação, que foram devidamente tratados. Após o processo de saneamento, os dados tornaram-se aptos para análises estatísticas e consultas analíticas, garantindo maior confiabilidade aos resultados obtidos.

# **Observação importante - delimitação de escopo:**
O conjunto de dados utilizado neste trabalho não contém informações no nível de restaurante, mas sim dados relacionados aos entregadores e ao contexto operacional das entregas.
Dessa forma, para manter a coerência técnica e metodológica da análise, as perguntas de negócio foram ajustadas para considerar a avaliação média por contexto urbano (cidade), que representa o nível máximo de agregação possível com consistência a partir dos dados disponíveis.

# **PERGUNTA 1:**
Como a avaliação média das entregas varia entre diferentes contextos urbanos?

```python
# Preparação
df_q1 = df.dropna(subset=['City', 'Delivery_person_Ratings'])
```

```python
# Avaliação média por cidade
avg_rating_by_city = (
    df_q1
    .groupby('City')['Delivery_person_Ratings']
    .mean()
    .reset_index()
    .sort_values(by='Delivery_person_Ratings', ascending=False)
)

avg_rating_by_city
```

A análise das avaliações médias das entregas evidencia diferenças relevantes entre os contextos urbanos analisados. As entregas realizadas em áreas classificadas como Urban apresentam a maior média de avaliação, seguidas pelas áreas Metropolitian. Em contrapartida, o contexto Semi-Urban registra a menor média entre os grupos observados.

Esses resultados sugerem que ambientes urbanos mais estruturados tendem a oferecer condições operacionais mais favoráveis, como melhor infraestrutura viária, menor distância média entre pontos de coleta e entrega e maior familiaridade dos entregadores com as rotas. Por outro lado, áreas semiurbanas podem enfrentar desafios logísticos adicionais que impactam negativamente a experiência do cliente.

Ainda que as diferenças nas avaliações não sejam extremas, elas indicam que o contexto urbano exerce influência na percepção de qualidade do serviço de entrega.

# **PERGUNTA 2:**
Existe relação entre o volume de entregas e a avaliação média do serviço nos diferentes contextos urbanos?

```python
# Preparação
df_q2 = df.dropna(subset=['City', 'Delivery_person_Ratings'])

volume_rating_by_city = (
    df_q2
    .groupby('City')
    .agg(
        total_entregas=('ID', 'count'),
        avaliacao_media=('Delivery_person_Ratings', 'mean')
    )
    .reset_index()
)

volume_rating_by_city
```

Ao analisar conjuntamente o volume de entregas e a avaliação média do serviço, observa-se que o contexto Metropolitian concentra a maior parte das entregas realizadas. No entanto, esse maior volume não se traduz na melhor avaliação média.

O contexto Urban, apesar de apresentar um número significativamente menor de entregas, registra a maior média de avaliações. Já o contexto Semi-Urban, que possui o menor volume operacional, também apresenta a menor avaliação média.

Esses achados indicam que não existe uma relação direta e proporcional entre volume de entregas e qualidade percebida do serviço. Ambientes com maior demanda podem enfrentar maior pressão operacional, o que tende a impactar a experiência do cliente, enquanto contextos menos complexos podem favorecer avaliações mais positivas.

# **PERGUNTA 3:**
O tempo médio de entrega varia de acordo com o contexto urbano?

```python
# Preparação:
df_q3 = df.dropna(subset=['City', 'Time_taken(min)'])
```

```python
# Tempo médio de entrega por contexto urbano
tempo_medio_por_cidade = (
    df_q3
    .groupby('City')['Time_taken(min)']
    .mean()
    .reset_index()
    .sort_values(by='Time_taken(min)')
)

tempo_medio_por_cidade
```

Os resultados demonstram diferenças expressivas no tempo médio de entrega entre os contextos urbanos analisados. As áreas Urban apresentam o menor tempo médio de entrega, enquanto o contexto Metropolitian registra valores intermediários. Em contraste, as áreas Semi-Urban apresentam um tempo médio significativamente mais elevado.

Esse comportamento pode ser explicado por fatores como maior densidade de estabelecimentos, melhor infraestrutura e menor distância média percorrida em áreas urbanas. Já em contextos semiurbanos, maiores deslocamentos e limitações logísticas tendem a impactar negativamente a eficiência das entregas.

Dessa forma, os resultados reforçam que o contexto urbano é um fator determinante para o desempenho operacional das entregas.

# **PERGUNTA 4:**
Condições de tráfego influenciam o tempo médio de entrega?

```python
# Preparação:
df_q4 = df.dropna(subset=['Road_traffic_density', 'Time_taken(min)'])
```

```python
# Tempo médio de entrega por condição de tráfego
tempo_medio_trafego = (
    df_q4
    .groupby('Road_traffic_density')['Time_taken(min)']
    .mean()
    .reset_index()
    .sort_values(by='Time_taken(min)')
)

tempo_medio_trafego
```

A análise evidencia uma relação clara entre a densidade do tráfego e o tempo médio de entrega. Em condições de tráfego Low, as entregas apresentam os menores tempos médios, enquanto cenários classificados como Jam registram os maiores tempos de entrega.

Observa-se um aumento gradual no tempo médio à medida que a densidade do tráfego se intensifica, passando por Medium e High até atingir os valores mais elevados em situações de congestionamento.

Esses resultados são coerentes com a dinâmica urbana, uma vez que o aumento do tráfego reduz a fluidez das rotas e impacta diretamente a eficiência operacional do serviço de delivery.

## **PERGUNTA 5:**
Condições climáticas impactam o tempo de entrega?

```python
# Preparação:
df_q5 = df.dropna(subset=['Weatherconditions', 'Time_taken(min)'])
```

```python
# Tempo médio de entrega por condição climática:
tempo_medio_clima = (
    df_q5
    .groupby('Weatherconditions')['Time_taken(min)']
    .mean()
    .reset_index()
    .sort_values(by='Time_taken(min)')
)

tempo_medio_clima
```

Os dados analisados indicam que as condições climáticas influenciam o desempenho das entregas. Situações classificadas como Sunny apresentam os menores tempos médios de entrega, refletindo um ambiente operacional mais favorável.

À medida que as condições climáticas se tornam mais adversas, observa-se um aumento no tempo médio de entrega. Condições como Stormy, Sandstorms e Windy apresentam tempos intermediários, enquanto Fog e Cloudy estão associadas aos maiores tempos observados.

Esse comportamento pode ser explicado pela redução da visibilidade, maior necessidade de cautela e impactos na mobilidade, fatores que afetam diretamente a velocidade de deslocamento dos entregadores.

# **PERGUNTA 6:**
Existe diferença no tempo de entrega de acordo com o tipo de veículo utilizado?

```python
# Preparação:
df_q6 = df.dropna(subset=['Type_of_vehicle', 'Time_taken(min)'])
```

```python
# Tempo médio de entrega por tipo de veículo:
tempo_medio_veiculo = (
    df_q6
    .groupby('Type_of_vehicle')['Time_taken(min)']
    .mean()
    .reset_index()
    .sort_values(by='Time_taken(min)')
)

tempo_medio_veiculo
```

A análise do tempo médio de entrega por tipo de veículo evidencia diferenças no desempenho operacional. Veículos mais leves e ágeis, como electric_scooter e scooter, apresentam os menores tempos médios de entrega, indicando maior facilidade de deslocamento em ambientes urbanos.

As entregas realizadas por bicycle apresentam tempos intermediários, possivelmente impactadas por limitações físicas, apesar da vantagem de circulação em áreas restritas. Já o motorcycle registra o maior tempo médio entre os tipos analisados, o que pode estar relacionado à maior dependência das condições de tráfego e ao perfil das rotas atendidas.

Esses resultados demonstram que o tipo de veículo é um fator relevante para a eficiência operacional do serviço de delivery.

# **PERGUNTA 7:**
Entregas realizadas durante períodos festivos apresentam diferença de desempenho em relação a períodos regulares?

```python
# Preparação:
df_q7 = df.dropna(subset=['Festival', 'Time_taken(min)'])
```

```python
# Tempo médio de entrega em períodos festivos vs. não festivos:
tempo_medio_festival = (
    df_q7
    .groupby('Festival')['Time_taken(min)']
    .mean()
    .reset_index()
)

tempo_medio_festival
```

Os resultados apontam uma diferença significativa no desempenho das entregas entre períodos festivos e não festivos. Durante períodos classificados como festivos, o tempo médio de entrega é substancialmente maior quando comparado aos períodos regulares.

Esse aumento pode ser atribuído a fatores como maior volume de pedidos, intensificação do tráfego e maior complexidade logística associada a datas comemorativas. Tais condições impõem desafios adicionais à operação, impactando diretamente a eficiência das entregas.

Dessa forma, os períodos festivos se mostram como cenários críticos do ponto de vista operacional, demandando planejamento específico para mitigar atrasos e preservar a qualidade do serviço.
